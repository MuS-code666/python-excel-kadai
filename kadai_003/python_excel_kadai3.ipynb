```python
import pandas as pd
from openpyxl.styles import PatternFill

# 1. 各年のデータをSheet1から読み込み、データフレームを作成
# 各ファイルには「売上年」「商品」「金額（千円）」の列が含まれている
df_2022 = pd.read_excel("2022_年間売上表.xlsx", sheet_name="Sheet1")
df_2023 = pd.read_excel("2023_年間売上表.xlsx", sheet_name="Sheet1")

# データを連結
df_all = pd.concat([df_2022, df_2023], ignore_index=True)

# 2. 「商品」「売上年」ごとにグループ化し、合計金額に集約
df_summary = (
    df_all.groupby(["商品", "売上年"], as_index=False)["金額（千円）"]
    .sum()
    .rename(columns={"金額（千円）": "合計金額"})
)

# 3. 新規Excelファイルに書き込み
output_file = "売上集計表.xlsx"
df_summary.to_excel(output_file, index=False, sheet_name="Sheet1")

# 4. ヘッダー行（1行目）を薄いグレー（#F2F2F2）に設定して保存
from openpyxl import load_workbook

wb = load_workbook(output_file)
ws = wb["Sheet1"]

header_fill = PatternFill(start_color="F2F2F2", end_color="F2F2F2", fill_type="solid")
for cell in ws[1]:
    cell.fill = header_fill

wb.save(output_file)

print(df_summary)
```
