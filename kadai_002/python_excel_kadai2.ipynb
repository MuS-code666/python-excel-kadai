```python
import pandas as pd

# 1. データフレームの作成
data = {
    "日付": ["2023-05-17", "2023-05-18", "2023-05-19", "2023-05-20", "2023-05-21"],
    "社員名": ["山田", "佐藤", "鈴木", "田中", "高橋"],
    "売上": [100, 200, 150, 300, 250],
    "部門": ["メーカー", "代理店", "メーカー", "商社", "代理店"],
}
df = pd.DataFrame(data)
df["日付"] = pd.to_datetime(df["日付"])

# 2. 「平均売上」列を作成（全体の売上の平均）
avg_sales = df["売上"].mean()
df["平均売上"] = avg_sales


# 3. 「業績ランク」列を作成
def performance(row):
    if row["売上"] >= row["平均売上"] + 50:
        return "A"
    elif row["売上"] >= row["平均売上"]:
        return "B"
    else:
        return "C"


df["業績ランク"] = df.apply(performance, axis=1)

print(df)

# 4. 新規ファイル「業績.xlsx」に書き出し
df.to_excel("業績.xlsx", index=False)
```
